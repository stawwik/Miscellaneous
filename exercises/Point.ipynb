# Class making exercise

- The **Point** class must accept 3 values on initialization (x, y, and z) and
have x, y, and z attributes.
- It must also have a helpful string representation.
- Additionally, point objects **should be comparable to each other** (two points are equal if their coordinates are the same and not equal otherwise).
- Allow **Point** objects to be **added** and **subtracted** from each other.
- Allow **Point** objects to be **scaled** up and down by numbers.
- Allow Point objects to be **unpacked using multiple assignment**.

```python
class Point:
	"""Class representing a three-dimensional point."""

	def __init__(self, x:float, y: float, z: float):
		self.x, self.y, self.z = x, y, z

	def __repr__(self):
		"""Return dev-readable representation of Point."""
		x, y, z = self
		return f'Point({x=}, {y=}, {z=})'

	def __eq__(self, other):
		"""Return True if our point is equal to the other point."""
		if not isinstance(other, Point):
			return NotImplemented

		x1, y1, z1 = self
		x2, y2, z2 = other

		return (x1, y1, z1) == (x2, y2, z2)

	def __add__(self, other):
		"""Return copy of our point, shifted by other."""
		if not isinstance(other, Point):
			return NotImplemented
		x1, y1, z1 = self
		x2, y2, z2 = other
		return Point(x1 + x2, y1 + y2, z1 + z2)

	def __sub__(self, other):
		"""Return copy of our point, shifted by other."""
		if not isinstance(other, Point):
			return NotImplemented
		x1, y1, z1 = self
		x2, y2, z2 = other
		return Point(x1 - x2, y1 - y2, z1 - z2)

	def __mul__(self, scalar: float):
		"""Return new copy of our point, scaled by given value."""
		x, y, z = self
		return Point(scalar * x, scalar * y, scalar * z)

	def __rmul__(self, scalar):
		return self.__mul__(scalar)

	def __iter__(self):
		yield from (self.x, self.y, self.z)
```
